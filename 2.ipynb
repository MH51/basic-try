```python
"""
        *
       **
      ***
       **
        *
"""
#x=int(input("enter  number"))
x=10
for i in range(0,x):
    if i<5:
        for j in range(0,x-i-6):
            print(" ",end=" ")
        print("* "*(i+1))
    else:
        for j in range(4,i):
            print(" ",end=" ")
        print("* "*(x-i-1))
```
